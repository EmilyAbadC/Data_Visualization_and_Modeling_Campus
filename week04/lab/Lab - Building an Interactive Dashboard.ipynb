## 9. Create any calculated features

If the dashboard needs a calculated category, date component, percentage, rate, or other calculated feature, create it here. If no calculated feature is needed, leave the code cell blank and explain why.



```python
def group_genre(genre):
    if "Pop" in genre:
        return "Pop"
    elif "Hip-Hop" in genre or "R&B" in genre:
        return "Hip-Hop/R&B"
    elif "Country" in genre:
        return "Country"
    elif genre in ["Indie Rock", "Alternative"]:
        return "Rock/Alternative"
    else:
        return "Other"

analysis["genre_group"] = analysis["primary_genre"].apply(group_genre)

analysis["genre_group"].value_counts()
```

**What calculated feature or cleaned label did you create, and how will it support the dashboard?**
I created a genre_group feature to combine similar genre labels into broader categories. The original data contains several mixed or closely related genre labels, so grouping them makes the interactive comparisons easier to understand and avoids having too many very small categories.

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def create_figures(selected_genre):
    if selected_genre == "All":
        filtered = analysis
    else:
        filtered = analysis[analysis["genre_group"] == selected_genre]

    fig1 = px.scatter(
        filtered,
        x="danceability",
        y="streams_2025_billions",
        color="genre_group",
        hover_name="song_title",
        title="Danceability and Streaming Popularity",
        labels={
            "danceability": "Danceability",
            "streams_2025_billions": "Streams in 2025 (Billions)",
            "genre_group": "Genre"
        }
    )

    fig1.update_xaxes(range=[0, 1])
    fig1.update_yaxes(range=[0, 1.8])

    fig2 = px.scatter(
        filtered,
        x="energy",
        y="streams_2025_billions",
        color="genre_group",
        hover_name="song_title",
        title="Energy and Streaming Popularity",
        labels={
            "energy": "Energy",
            "streams_2025_billions": "Streams in 2025 (Billions)",
            "genre_group": "Genre"
        }
    )

    fig2.update_xaxes(range=[0, 1])
    fig2.update_yaxes(range=[0, 1.8])

    return fig1, fig2

```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("Spotify Wrapped 2025: What Makes a Top Song?"),

    html.P(
        "Explore how danceability and energy relate to streaming popularity "
        "among Spotify's top 50 songs of 2025. Use the genre filter to compare "
        "patterns across different genre groups."
    ),

    html.Label("Select a genre:"),

    dcc.Dropdown(
        id="genre-dropdown",
        options=genre_options,
        value="All",
        clearable=False
    ),

    dcc.Graph(id="danceability-graph"),

    dcc.Graph(id="energy-graph"),

    html.H3("Conclusion"),
    html.P(
        "The dashboard helps compare how danceability and energy relate to "
        "streaming popularity across different genre groups."
    ),

    html.H3("Limitation"),
    html.P(
        "The dataset only includes Spotify's top 50 songs of 2025, so the "
        "patterns should not be generalized to all songs on Spotify."
    )
])

```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python
@app.callback(
    Output("danceability-graph", "figure"),
    Output("energy-graph", "figure"),
    Input("genre-dropdown", "value")
)
def update_dashboard(selected_genre):
    fig1, fig2 = create_figures(selected_genre)
    return fig1, fig2

```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
pop_fig1, pop_fig2 = create_figures("Pop")
country_fig1, country_fig2 = create_figures("Country")

print(type(pop_fig1))
print(type(pop_fig2))
print(type(country_fig1))
print(type(country_fig2))

```
